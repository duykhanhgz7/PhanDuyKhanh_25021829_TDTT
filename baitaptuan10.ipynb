```python
def canatob(a, b):
    c = list(a)
    for i in a:
        if i == i.lower():
            c.remove(i)
    if ''.join(c) == b:
        return True
    else:
        a = a.upper()
        k = 0
        cnt = 0
        for i in b:
            for j in range(k, len(a)):
                if a[j] == i:
                    cnt += 1
                    k = j + 1
                    break
        if cnt == len(b):
            return True
        return False

n = int(input())
for i in range(n):
    a = input()
    b = input()
    if canatob(a, b):
        print('YES')
    else:
        print('NO')
                    
```

```python
# 6
def nextnumber(arr):
    a = int(''.join(arr))
    k = ''
    for i in range(len(arr)-2, -1, -1):
        for j in range(len(arr)-1, i, -1):
            ta = arr.copy()
            ta[i], ta[j] = ta[j], ta[i]
            b = int(''.join(ta))
            if b > a:
                arr[i], arr[j] = arr[j], arr[i]
                tail = sorted(arr[i+1:])
                k = ' '.join(arr[0:i+1]) + ' ' + ' '.join(tail)
                break
        if k != '':
            break
    return k
N = input()
arr = list(map(str, input().split()))
print(nextnumber(arr))
```

```python
# 7
def canPlaceFlowers(flowerbed, k):
    if k == 0:
        return 0
    for i in range(len(flowerbed)):
        if flowerbed[i] == '0':
            a = flowerbed.copy()
            a[i] = '1'
            b = ''.join(a)
            if '11' in b:
                continue
            else:
                c = list(a)
                return 1 + canPlaceFlowers(c, k - 1)
    return 0
arr = list(map(str, input().split()))
n = int(input())
if canPlaceFlowers(arr, n) == n:
    print('True')
else:
    print('False')
```

```python
# 8
def is_path_crossing(moves):
    phi = 0
    start = [0,0]
    gps = [[0,0]]
    for i in moves:
        if i == 'R':
            phi -= 90
        if i == 'L':
            phi += 90
        if i == 'D':
            phi -= 180
        if i == 'U':
            phi += 0
        if (phi//90) % 4 == 0:
            start[0] += 1
            ns = start.copy()
            gps.append(ns)
        if (phi//90) % 4 == 1 or (phi//90) % 4 == -3:
            start[1] += 1
            ns = start.copy()
            gps.append(ns)
        if abs((phi//90) % 4) == 2:
            start[0] -= 1
            ns = start.copy()
            gps.append(ns)
        if (phi//90)% 4 == 3 or (phi//90) % 4 == -1:
            start[1] -= 1
            ns = start.copy()
            gps.append(ns)
    
    for i in gps:
        if gps.count(i) >= 2:
            return True
    return False

move = input()
print(is_path_crossing(move))
```

```python
# 10
def findRadius(houses, heaters):
    a = min(houses)
    b = max(houses)
    maxdis = 0
    for i in heaters:
        h = abs(i - a)
        k = abs(i - b)
        if h + k > maxdis:
            maxdis = h + k
    mindis = 0
    for i in heaters:
        h = abs(i - a)
        k = abs(i - b)
        if h + k <= maxdis:
            mindis = h + k
            maxdis = mindis
    for i in heaters:
        h = abs(i - a)
        k = abs(i - b)
        if h + k == mindis:
            if h > k:
                return h
            else:
                return k

house = list(map(int, input().split()))
heater = list(map(int, input().split()))
print(findRadius(house, heater))
```
