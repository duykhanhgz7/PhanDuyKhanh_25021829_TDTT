```python
#bai 2
n = 0
for i in range(100):
    n += 1
    if n % 6 == 0:
        print(n, end = ' ')
    if n == 100:
        break
```

```python
h = int(input("Nhập h: "))

for i in range(1, h):
    print(" " * (h - i) + "* " * i)
print("*" * (2*h - 1))
```
