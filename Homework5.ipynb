```python
matrix = [
    [1, 22, 333],
    [444, 5, 66],
    [7, 88, 9]
]

# Giả sử ta muốn mỗi cột rộng 5 khoảng trắng
for row in matrix:
    for val in row:
        print(f"{val:<5}", end="") # "<5" là căn lề trái, độ rộng 5
    print() # Xuống dòng sau mỗi hàng
```

```python
# Bước 1: Nhập số hàng n và số cột m
n, m = map(int, input("Nhập n và m: ").split())

# Bước 2: Nhập các phần tử của ma trận
matrix = []
print(f"Nhập các phần tử cho ma trận {n}x{m}:")
for i in range(n):
    # Nhập cả một dòng, chuyển thành list số nguyên
    row = list(map(int, input().split()))
    matrix.append(row)

# Bước 3: In ma trận theo đúng định dạng
print("\nKết quả ma trận:")
for row in matrix:
    for val in row:
        # Sử dụng f-string với định dạng :>4
        # > : Căn lề phải
        # 4 : Độ rộng là 4 ký tự
        print(f"{val:>4}", end="")
    
    # Sau khi in xong một hàng thì xuống dòng
    print()
```
